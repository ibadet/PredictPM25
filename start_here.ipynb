### 1. Project Overview

This project predicts PM2.5 levels and compares the manual ARIMA/SARIMAX models with transformer models.
Main objectives:
- Explore PM2.5 trends
- Build forecasting models
- Evaluate prediction accuracy

### 2.Setup Instructions

Clone the repository and install required libraries:

pip install -r requirements.txt

> **Note**: The raw PurpleAir sensor data (California-wide) is too large to include in this repository.
> The pipeline starts from the pre-processed LA daily file (`processed_data/la_daily.csv`).
> To reproduce from scratch, download the raw data from [PurpleAir](https://www2.purpleair.com/) 
> and run `pm25_LA_preprocessing.ipynb` first. Otherwise, start directly from `pm25_data_cleaning.ipynb`.

### 3. Run the Notebooks in This Order


1. [PM2.5 LA Preprocessing](project_files/scripts/pm25_LA_preprocessing.ipynb) | ⚠️ Optional — raw CA data not included; start here only if you have the raw PurpleAir data |

   
3. [PM2.5 Data Cleaning](project_files/scripts/pm25_data_cleaning.ipynb) | **Start here** — LA daily data cleaning and quality checks |

4. [Exploratory Data Analysis](project_files/scripts/exploratory_data_analysis.ipynb)

5. [PM2.5 SARIMAX](project_files/scripts/pm25_sarimax.ipynb)

6. [PM2.5 Transformer](project_files/scripts/pm25_transformer.ipynb)
