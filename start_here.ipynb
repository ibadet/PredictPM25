### 1. Project Overview

This project predicts PM2.5 levels and compares the manual ARIMA/SARIMAX models with transformer models.
Main objectives:
- Explore PM2.5 trends
- Build forecasting models
- Evaluate prediction accuracy

### 2.Setup Instructions

Clone the repository and install required libraries:

pip install -r requirements.txt

### 3. Run the Notebooks in This Order


1. [PM2.5 LA Preprocessing](project_files/scripts/pm25_LA_preprocessing.ipynb)
   
2. [PM2.5 Data Cleaning](project_files/scripts/pm25_data_cleaning.ipynb)

3. [Exploratory Data Analysis](project_files/scripts/exploratory_data_analysis.ipynb)

4. [PM2.5 SARIMAX](project_files/scripts/pm25_sarimax.ipynb)

5. [PM2.5 Transformer](project_files/scripts/pm25_transformer.ipynb)
